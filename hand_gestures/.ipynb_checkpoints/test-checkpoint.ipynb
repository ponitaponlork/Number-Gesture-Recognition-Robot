```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model

# -------------------- MEDIAPIPE TASKS IMPORT --------------------
from mediapipe.tasks.python.vision.hand_landmarker import HandLandmarker, HandLandmarkerOptions
from mediapipe.tasks.python.core.base_options import BaseOptions
from mediapipe.tasks.python.vision.core.vision_task_running_mode import VisionTaskRunningMode

# Universal image wrapper (works on all 0.10.x versions)
import mediapipe as mp
mp_Image = mp.Image
ImageFormat = mp.ImageFormat

# -------------------- SETTINGS --------------------
MODEL_PATH = "best_gesture_model.h5"
HAND_MODEL_PATH = "hand_landmarker.task"
NUM_CLASSES = 6
# ---------------------------------------------------

# Load gesture model
model = load_model(MODEL_PATH)

# Create hand detector
options = HandLandmarkerOptions(
    base_options=BaseOptions(model_asset_path=HAND_MODEL_PATH),
    running_mode=VisionTaskRunningMode.VIDEO,
    num_hands=1
)

detector = HandLandmarker.create_from_options(options)

# Start webcam
cap = cv2.VideoCapture(0)
print("Press 'q' to quit.")

frame_i = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame = cv2.flip(frame, 1)
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # Convert OpenCV → MediaPipe Image
    mp_image = mp_Image(image_format=ImageFormat.SRGB, data=rgb)

    # Detect hands
    result = detector.detect_for_video(mp_image, frame_i)
    frame_i += 1

    prediction_text = "No hand detected"

    if result.hand_landmarks:
        lm_list = result.hand_landmarks[0]

        # Extract 63 landmark coordinates
        coords = []
        for lm in lm_list:
            coords.extend([lm.x, lm.y, lm.z])
        coords = np.array(coords).reshape(1, -1)

        # Predict
        pred = model.predict(coords, verbose=0)
        gesture = np.argmax(pred)
        confidence = np.max(pred)

        prediction_text = f"Gesture: {gesture} ({confidence*100:.1f}%)"

        # Draw dots
        for lm in lm_list:
            x = int(lm.x * frame.shape[1])
            y = int(lm.y * frame.shape[0])
            cv2.circle(frame, (x, y), 5, (0, 255, 0), -1)

    # Display text
    cv2.putText(frame, prediction_text, (10, 40),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

    cv2.imshow("Gesture Recognition", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import tf2onnx

# Load Keras model
model = load_model("best_gesture_model.h5")

# ---- FIX FOR TF 2.12+ ----
# tf2onnx expects this attribute but TF removed it
if not hasattr(model, "output_names"):
    model.output_names = ["output"]
# --------------------------

# Define model input signature
spec = (tf.TensorSpec((1, 63), tf.float32, name="input"),)

# Convert to ONNX
output_path = "gesture_model.onnx"
model_proto, _ = tf2onnx.convert.from_keras(
    model,
    input_signature=spec,
    output_path=output_path
)

print("ONNX model saved successfully:", output_path)
```
